```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import statsmodels.formula.api as sm

players = pd.read_csv('./datasets/mod_players.csv',low_memory=False)
```

#### The difference between correlation and regression (in the simplest terms) is that correlation analysis shows that there is a relationship or not between two variables, whereas regression shows is the extent of the relationship eg the correlation between wlr(win_percentage) and py_exp(pythogorean_expectation) is **0.945(94.5%)** whereas regression analysis shows that **an increase of 1% in py_exp increases the wlr by 0.876(87.6%)**

```python
players['wlr'].corr(players['py_exp'])
```

```python
reg1 = sm.ols(formula='wlr ~ py_exp',data=players).fit()
reg1.summary()
```
